# 1. Dataset Understanding

## What is a Dataset?

A dataset is a collection of information organized into rows and columns. Each row represents a record, and each column represents an attribute.

## What is a Feature?

A feature is an individual attribute or column in a dataset. Examples include customer tenure, monthly charges, and contract type.

## What is a Target Variable?

A target variable is the value we want to predict using a Machine Learning model. In our Customer Churn dataset, the target variable is `Churn`.

## Independent and Dependent Variables

* **Independent variables:** Input features used to predict an outcome, such as tenure and monthly charges.
* **Dependent variable:** The outcome we want to predict, which is `Churn` in this dataset.

## Numerical Features

Numerical features contain numeric values that represent quantities. Examples include `tenure` and `MonthlyCharges`.

## Categorical Features

Categorical features represent groups or categories. Examples include `gender`, `Contract`, and `PaymentMethod`.

## Discrete and Continuous Features

* **Discrete features:** Countable values, such as the number of dependents.
* **Continuous features:** Measurements that can take different values within a range, such as height or distance.

In our dataset, `tenure` is recorded as a whole number of months.

## Identifier Columns

An identifier is used to distinguish individual records. The `customerID` column identifies each customer. It is generally not useful as a direct predictive feature.

## Metadata

Metadata describes the dataset, including its name, source, domain, dimensions, and column information.

## Dataset Dimensions

Dataset dimensions refer to the number of rows and columns. Rows represent customer records, while columns represent customer attributes.

## Dataset Information

* **Dataset Name:** Telco Customer Churn
* **Domain:** Telecommunications
* **Dataset Source:** IBM Telco Customer Churn dataset
* **Target Variable:** `Churn`
* **Business Objective:** Understand customer churn and identify patterns associated with customers leaving the company.

## AI/ML Use Case

We can use features such as tenure, monthly charges, and contract type to predict the target variable `Churn`.

Before building a model, we must inspect the data and identify missing values, duplicates, unusual values, and relationships between features.

**Important:** This is an illustrative example. We must analyze the actual dataset before drawing conclusions about customer behavior.


```python
# Import the required libraries
import pandas as pd
from pathlib import Path

# Load the dataset
file_path = Path("dataset/Telco-Customer-Churn.csv")
df = pd.read_csv(file_path)

# Display the first five records
print("First Five Records:")
display(df.head())

# Display dataset dimensions
print("\nDataset Dimensions:")
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

# Display column names
print("\nColumn Names:")
print(df.columns.tolist())

# Display data types
print("\nData Types:")
print(df.dtypes)

# Identify numerical and categorical columns
numerical_columns = df.select_dtypes(
    include="number"
).columns.tolist()

categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("\nNumerical Columns:")
print(numerical_columns)

print("\nCategorical Columns:")
print(categorical_columns)

# Identify the target variable
print("\nTarget Variable: Churn")

print("\nCustomer Churn Counts:")
print(df["Churn"].value_counts(dropna=False))
```
