# Notebook 3 – Data Quality Assessment

## Step 1: Understand Concept

### 1. What Is Data Quality Assessment?

Data quality assessment is the process of checking a dataset for missing values, duplicate records, incorrect values, and inconsistencies.

### 2. Why Is It Important?

* Missing values can affect analysis and model performance.
* Duplicate records can lead to incorrect conclusions.
* Invalid values can produce misleading results.
* Inconsistent categories can create problems during data analysis.
* Checking data quality helps us prepare reliable data for EDA and machine learning.

### 3. Important Data Quality Checks

| Check                    | Purpose                                     | Pandas Function      |
| ------------------------ | ------------------------------------------- | -------------------- |
| Missing values           | Find empty or missing entries               | `isnull()`           |
| Missing-value counts     | Count missing entries per column            | `isnull().sum()`     |
| Missing-value percentage | Calculate the percentage of missing entries | `isnull().mean()`    |
| Duplicate rows           | Find repeated records                       | `duplicated()`       |
| Duplicate count          | Count repeated records                      | `duplicated().sum()` |
| Unique values            | Examine values in a column                  | `unique()`           |
| Value frequencies        | Count each category                         | `value_counts()`     |
| Data types               | Identify the type of each column            | `dtypes`             |

### 4. Common Data Quality Problems

* Missing values
* Duplicate records
* Invalid numerical values
* Inconsistent categorical values
* Unexpected values in important columns

### 5. Dataset

We will use the Telco Customer Churn dataset to identify potential data quality issues before analyzing customer behavior.

**Important:** We will inspect the data first. We will not automatically delete or modify records because each change must be justified.


## Step 2: Demonstrate Concept

### Real-World Example: Customer Data Quality

Imagine a telecommunications company stores customer information.

| Customer ID | Tenure | Monthly Charges | Contract       |
| ----------- | -----: | --------------: | -------------- |
| C001        |     12 |            70.5 | Month-to-month |
| C002        |     24 |         Missing | One year       |
| C002        |     24 |         Missing | One year       |
| C003        |     -5 |            60.0 | Two year       |

This illustrative dataset contains several potential problems:

1. **Missing value:** Customer C002 has no monthly charge recorded.
2. **Duplicate record:** Customer C002 appears twice.
3. **Invalid value:** A tenure of -5 months is suspicious and needs investigation.
4. **Data reliability:** Incorrect or incomplete records may affect business decisions.

### How Can We Identify These Problems?

* Use `isnull()` to find missing values.
* Use `duplicated()` to find duplicate records.
* Use `describe()` to examine numerical ranges.
* Use `unique()` and `value_counts()` to inspect categorical values.

**Expected outcome:** We will identify possible data quality issues and decide which ones require further investigation before cleaning the dataset.


```python
# Import the required library
import pandas as pd
from pathlib import Path

# Load the Telco Customer Churn dataset
file_path = Path("dataset/Telco-Customer-Churn.csv")
df = pd.read_csv(file_path)

# 1. Check missing values in each column
print("Missing Values Per Column:")
print(df.isnull().sum())

# 2. Calculate missing-value percentages
print("\nMissing Value Percentage:")
missing_percentage = (df.isnull().mean() * 100).round(2)
print(missing_percentage)

# 3. Display columns containing missing values
print("\nColumns With Missing Values:")
print(missing_percentage[missing_percentage > 0])

# 4. Check duplicate rows
print("\nNumber of Duplicate Rows:")
print(df.duplicated().sum())

# 5. Display duplicate rows, if any
print("\nDuplicate Records:")
display(df[df.duplicated()])

# 6. Examine unique values in important categorical columns
categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("\nUnique Values in Categorical Columns:")
for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].value_counts(dropna=False).head(20))

# 7. Examine numerical statistics for suspicious values
print("\nNumerical Column Statistics:")
display(df.describe().T)

# 8. Check the tenure column for negative values
if "tenure" in df.columns:
    print("\nNegative Tenure Records:")
    display(df[df["tenure"] < 0])

# 9. Check the MonthlyCharges column for negative values
if "MonthlyCharges" in df.columns:
    print("\nNegative Monthly Charges:")
    display(df[df["MonthlyCharges"] < 0])

# 10. Check the TotalCharges column for invalid or missing values
if "TotalCharges" in df.columns:
    total_charges = pd.to_numeric(df["TotalCharges"], errors="coerce")

    print("\nTotalCharges Values That Could Not Be Converted:")
    print(total_charges.isnull().sum())

    print("\nNegative TotalCharges Records:")
    display(df[total_charges < 0])

# 11. Summarize dataset quality
print("\nDataset Quality Summary:")
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])
print("Total Missing Values:", df.isnull().sum().sum())
print("Number of Duplicate Rows:", df.duplicated().sum())
```
