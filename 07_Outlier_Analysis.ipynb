# Outlier Analysis

## What is an Outlier?

An outlier is a data point that is significantly different from most other observations in a dataset.

Example:
If most customers have monthly charges between 20 and 100, but one customer has a monthly charge of 500, that value may be an outlier. However, we must check whether the value is actually incorrect before making any changes.

## 1. Why Do Outliers Occur?

Outliers can occur because of:

- Data-entry errors.
- Incorrect measurements.
- Equipment or system errors.
- Unusual customer behavior.
- Genuine extreme observations.
- Natural variation in the data.

## 2. Types of Outliers

### Univariate Outliers

An unusual value in one feature when that feature is examined individually.

Example:
A customer has an unusually high MonthlyCharges value compared with other customers.

### Multivariate Outliers

An observation that is unusual when two or more features are considered together, even if its individual values are not extreme.

Example:
A customer has a combination of unusually low tenure and unusually high TotalCharges compared with the general pattern in the dataset.

## 3. IQR Method

IQR stands for Interquartile Range.

IQR = Q3 - Q1

- Q1: 25th percentile.
- Q3: 75th percentile.
- IQR: Difference between Q3 and Q1.

Lower Bound = Q1 - 1.5 × IQR

Upper Bound = Q3 + 1.5 × IQR

Values outside these bounds are flagged as potential outliers.

The IQR method is useful because it is less affected by extreme values than the mean and standard deviation.

## 4. Z-Score Method

A Z-score measures how many standard deviations a value is away from the mean.

Formula:

Z = (Value - Mean) / Standard Deviation

A common rule flags values with an absolute Z-score greater than 3 as potential outliers.

This method works best when the numerical feature has a reasonably normal distribution. Skewed data can produce misleading flags.

## 5. Box Plot

A box plot displays:

- Median.
- First quartile (Q1).
- Third quartile (Q3).
- Interquartile range.
- Potential outliers beyond the whiskers.

Box plots help us identify unusual values visually.

## 6. Scatter Plot

A scatter plot displays the relationship between two numerical features.

It can help identify observations that appear unusual compared with the overall relationship between the features.

## 7. How Should We Handle Outliers?

For every detected outlier, consider these questions:

1. Is it a genuine observation?
2. Is it a data-entry or measurement error?
3. Is it an unusual but valid observation?
4. Should it be removed?
5. Should it be transformed?
6. Should it be retained?

### Possible Decisions

- Remove: Only when the value is confirmed to be invalid or removal is otherwise justified.
- Transform: Consider transformations such as logarithms when the distribution is highly skewed and the transformation suits the analysis.
- Retain: Keep valid observations that represent genuine customer behavior.
- Investigate: Check suspicious values against business rules or the original data source.

Important: Do not automatically remove every outlier. Detection identifies unusual observations; it does not prove that they are errors.

## Why Is Outlier Analysis Important in AI/ML?

- Helps identify data-quality problems.
- Helps us understand unusual observations.
- Some algorithms are sensitive to extreme values.
- Supports informed preprocessing decisions.
- Helps prevent useful information from being discarded unnecessarily.

# Outlier Analysis in Customer Churn

## Business Scenario

A telecommunications company wants to analyze customer behavior using the Telco Customer Churn dataset.

Some customers may have unusually high monthly charges, long tenure, or high total charges.

We need to determine whether these observations are errors or genuine customer records.

## Analysis 1: Univariate Outliers Using IQR

We will examine:

- tenure
- MonthlyCharges
- TotalCharges

For each numerical feature, we will calculate Q1, Q3, IQR, lower bound, and upper bound.

Values outside the bounds will be flagged as potential outliers.

## Analysis 2: Outliers Using Z-Score

We will calculate Z-scores for the numerical features.

Values with an absolute Z-score greater than 3 will be flagged as potential outliers.

We will compare the results with the IQR method because the two methods may identify different observations.

## Analysis 3: Box Plots

We will use box plots to visualize the distributions of the numerical features.

The plots will help us identify potential outliers and understand the spread of the data.

## Analysis 4: Multivariate Outliers Using a Scatter Plot

We will examine tenure against TotalCharges.

This helps us investigate customers whose combinations of tenure and total charges differ from the general pattern.

An unusual point is not necessarily incorrect. It may represent a genuine customer record.

## Analysis 5: Documenting Outlier Decisions

For each numerical feature, we will record:

- Number of potential outliers.
- Percentage of observations flagged.
- Whether the values are valid or require further investigation.
- Whether removal, transformation, or retention is appropriate.
- The reasoning behind the decision.

The notebook will not automatically remove or transform any observations.

## Expected Outcome

By the end of this analysis, we should understand how to:

- Detect potential univariate outliers.
- Investigate multivariate outliers visually.
- Compare the IQR and Z-score methods.
- Distinguish unusual observations from confirmed data errors.
- Document a reasoned decision for handling outliers.

```python
# Step 3: Implement the Concept – Outlier Analysis

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Load the dataset
file_path = Path("dataset/Telco-Customer-Churn.csv")

df = pd.read_csv(file_path)

# Convert TotalCharges to numeric
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

# Select numerical features
numerical_features = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

# Prepare numerical data
numeric_df = df[numerical_features].copy()

print("Dataset shape:", df.shape)
print("\nNumerical features:", numerical_features)
print("\nMissing values in numerical features:")
print(numeric_df.isnull().sum())


# 1. Detect Univariate Outliers Using IQR
print("\n" + "=" * 60)
print("1. IQR METHOD")
print("=" * 60)

iqr_results = {}

for feature in numerical_features:

    values = numeric_df[feature].dropna()

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_mask = (
        (numeric_df[feature] < lower_bound)
        | (numeric_df[feature] > upper_bound)
    )

    outlier_count = int(outlier_mask.sum())
    valid_count = int(numeric_df[feature].notna().sum())
    outlier_percentage = (
        outlier_count / valid_count * 100
        if valid_count > 0 else 0
    )

    iqr_results[feature] = {
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": outlier_count,
        "Outlier Percentage": outlier_percentage
    }

    print(f"\nFeature: {feature}")
    print(f"Q1: {q1:.2f}")
    print(f"Q3: {q3:.2f}")
    print(f"IQR: {iqr:.2f}")
    print(f"Lower Bound: {lower_bound:.2f}")
    print(f"Upper Bound: {upper_bound:.2f}")
    print(f"Potential Outliers: {outlier_count}")
    print(f"Outlier Percentage: {outlier_percentage:.2f}%")


# 2. Detect Potential Outliers Using Z-Score
print("\n" + "=" * 60)
print("2. Z-SCORE METHOD")
print("=" * 60)

zscore_results = {}

for feature in numerical_features:

    values = numeric_df[feature]

    mean_value = values.mean()
    std_value = values.std(ddof=0)

    if pd.isna(std_value) or std_value == 0:
        z_scores = pd.Series(
            np.nan,
            index=values.index
        )

        outlier_mask = pd.Series(
            False,
            index=values.index
        )

    else:
        z_scores = (values - mean_value) / std_value
        outlier_mask = z_scores.abs() > 3

    outlier_count = int(outlier_mask.sum())
    valid_count = int(values.notna().sum())

    outlier_percentage = (
        outlier_count / valid_count * 100
        if valid_count > 0 else 0
    )

    zscore_results[feature] = {
        "Mean": mean_value,
        "Standard Deviation": std_value,
        "Outlier Count": outlier_count,
        "Outlier Percentage": outlier_percentage
    }

    print(f"\nFeature: {feature}")
    print(f"Mean: {mean_value:.2f}")
    print(f"Standard Deviation: {std_value:.2f}")
    print(f"Potential Outliers (|Z| > 3): {outlier_count}")
    print(f"Outlier Percentage: {outlier_percentage:.2f}%")


# 3. Compare IQR and Z-Score Results
print("\n" + "=" * 60)
print("3. COMPARISON OF DETECTION METHODS")
print("=" * 60)

comparison = pd.DataFrame({
    "IQR Outlier Count": {
        feature: iqr_results[feature]["Outlier Count"]
        for feature in numerical_features
    },
    "Z-Score Outlier Count": {
        feature: zscore_results[feature]["Outlier Count"]
        for feature in numerical_features
    }
})

print(comparison)


# 4. Visualize Potential Outliers Using Box Plots
print("\n" + "=" * 60)
print("4. BOXPLOTS")
print("=" * 60)

for feature in numerical_features:

    values = numeric_df[feature].dropna()

    plt.figure(figsize=(8, 4))
    plt.boxplot(values, vert=False)

    plt.title(f"Box Plot of {feature}")
    plt.xlabel(feature)
    plt.tight_layout()
    plt.show()


# 5. Multivariate Outlier Investigation Using a Scatter Plot
print("\n" + "=" * 60)
print("5. MULTIVARIATE OUTLIER INVESTIGATION")
print("=" * 60)

scatter_df = df[
    ["tenure", "TotalCharges", "Churn", "customerID"]
].dropna(
    subset=["tenure", "TotalCharges"]
)

plt.figure(figsize=(9, 6))

for churn_status, group in scatter_df.groupby("Churn"):

    plt.scatter(
        group["tenure"],
        group["TotalCharges"],
        alpha=0.5,
        label=f"Churn = {churn_status}"
    )

plt.title("Tenure vs TotalCharges")
plt.xlabel("Tenure")
plt.ylabel("TotalCharges")
plt.legend()
plt.tight_layout()
plt.show()


# 6. List Potential IQR Outlier Records for Investigation
print("\n" + "=" * 60)
print("6. POTENTIAL OUTLIER RECORDS")
print("=" * 60)

for feature in numerical_features:

    lower_bound = iqr_results[feature]["Lower Bound"]
    upper_bound = iqr_results[feature]["Upper Bound"]

    outlier_mask = (
        (df[feature] < lower_bound)
        | (df[feature] > upper_bound)
    )

    outlier_records = df.loc[
        outlier_mask,
        ["customerID", feature, "tenure",
         "MonthlyCharges", "TotalCharges", "Churn"]
    ]

    print(f"\nFeature: {feature}")
    print("Sample potential outliers:")

    if outlier_records.empty:
        print("No potential outliers detected by IQR.")
    else:
        print(outlier_records.head(10).to_string(index=False))


# 7. Document Outlier Handling Decisions
print("\n" + "=" * 60)
print("7. OUTLIER HANDLING DECISIONS")
print("=" * 60)

decision_log = pd.DataFrame([
    {
        "Feature": feature,
        "Potential Outliers": iqr_results[feature]["Outlier Count"],
        "Initial Decision": "Investigate before changing",
        "Reasoning": (
            "Check source data and business context. "
            "Do not remove based on statistical thresholds alone."
        )
    }
    for feature in numerical_features
])

print(decision_log.to_string(index=False))


# 8. Final Summary
print("\n" + "=" * 60)
print("8. FINAL SUMMARY")
print("=" * 60)

print("Outlier analysis completed.")
print("IQR and Z-score methods were applied.")
print("Box plots and a scatter plot were generated.")
print("Potential outlier records were listed for investigation.")
print("A decision log was created to document the reasoning.")
print("\nNo records were automatically removed or transformed.")
```
