# Distribution Analysis

## What Is Distribution Analysis?

Distribution analysis helps us understand how numerical values are spread across a dataset.

It helps us identify the shape of the data, its center, its spread, its skewness, and the presence of potential outliers.

## 1. Normal Distribution

A normal distribution is a bell-shaped, symmetric distribution.

Characteristics:
- Mean, median, and mode are approximately equal.
- Most values are concentrated around the center.
- Fewer values appear toward the extremes.
- Skewness is approximately zero.

Example:
Measurements such as human heights may approximately follow a normal distribution in some populations.

## 2. Uniform Distribution

A uniform distribution occurs when values or outcomes have approximately equal probabilities across a defined range.

Characteristics:
- The probability is approximately equal across the range for a continuous uniform distribution.
- The distribution has a rectangular shape.
- The mean and median are equal for an ideal continuous uniform distribution.

Example:
A simulated random number generated uniformly between 0 and 1.

## 3. Symmetric Distribution

A symmetric distribution has approximately the same shape on both sides of its center.

Characteristics:
- The left and right sides look similar.
- Mean and median are often close.
- Skewness is approximately zero.

A symmetric distribution is not necessarily a normal distribution.

## 4. Skewness

Skewness measures the asymmetry of a distribution.

### Right-Skewed Distribution

A right-skewed distribution has a longer tail on the right side.

Characteristics:
- A few observations have unusually high values.
- The mean is often greater than the median.
- Skewness is positive.

Example:
Customer TotalCharges may be right-skewed because some customers have stayed with the company for a long time.

### Left-Skewed Distribution

A left-skewed distribution has a longer tail on the left side.

Characteristics:
- A few observations have unusually low values.
- The mean is often less than the median.
- Skewness is negative.

### Approximately Symmetric Distribution

When skewness is close to zero, the distribution may be approximately symmetric.

However, skewness alone does not prove that a distribution is symmetric or normal.

## 5. Kurtosis

Kurtosis describes aspects of the tails and extreme values of a distribution.

In pandas, the `kurt()` method returns excess kurtosis.

- Positive excess kurtosis: Heavier tails than a normal distribution, often indicating more extreme observations.
- Negative excess kurtosis: Lighter tails than a normal distribution.
- Excess kurtosis near zero: Similar kurtosis to a normal distribution.

Kurtosis does not directly measure whether the distribution is peaked, and it should not be used alone to identify outliers.

## 6. Shape of a Distribution

The shape describes how values are arranged.

Common shapes include:
- Bell-shaped.
- Uniform or rectangular.
- Right-skewed.
- Left-skewed.
- Bimodal, with two prominent peaks.
- Multimodal, with several prominent peaks.

## 7. Center of a Distribution

The center represents a typical value.

Common measures:
- Mean: Arithmetic average.
- Median: Middle value after sorting.
- Mode: Most frequent value.

The median is often more representative than the mean when a distribution is strongly skewed.

## 8. Spread of a Distribution

Spread describes how far values vary from one another.

Common measures:
- Range: Maximum minus minimum.
- Variance: Average squared deviation from the mean, using a specified variance convention.
- Standard deviation: Square root of the variance.
- IQR: Third quartile minus first quartile.

## 9. Presence of Potential Outliers

Potential outliers are observations that differ substantially from the general pattern.

We can investigate them using:
- Histograms.
- Box plots.
- IQR bounds.
- Z-scores.

An unusual value is not automatically an error.

## Why Is Distribution Analysis Important in AI/ML?

- Helps us understand the data before model training.
- Identifies skewed numerical features.
- Helps us select appropriate preprocessing methods.
- Highlights potential outliers.
- Helps us decide whether transformations may be useful.
- Improves our understanding of the assumptions behind some statistical methods and machine learning models.

# Distribution Analysis in Customer Churn

## Business Scenario

A telecommunications company wants to understand the numerical characteristics of its customers.

We will use the Telco Customer Churn dataset and analyze three important features:

- `tenure`: Number of months a customer has stayed with the company.
- `MonthlyCharges`: Monthly amount charged to a customer.
- `TotalCharges`: Total amount charged to a customer.

## Analysis 1: Distribution Shape

We will create histograms to understand how the values are distributed.

We will look for:
- Bell-shaped distributions.
- Approximately uniform distributions.
- Left-skewed distributions.
- Right-skewed distributions.
- Multiple peaks.

We will not assume that a feature follows a particular distribution before examining its plot.

## Analysis 2: Center

We will calculate:
- Mean.
- Median.
- Mode.

Comparing the mean and median helps us understand whether extreme values may be influencing the average.

## Analysis 3: Spread

We will calculate:
- Minimum and maximum.
- Range.
- Variance.
- Standard deviation.
- First and third quartiles.
- Interquartile range.

These statistics help us understand how widely customer values vary.

## Analysis 4: Skewness

We will calculate skewness for each numerical feature.

Interpretation:
- Positive skewness indicates a longer or heavier right tail.
- Negative skewness indicates a longer or heavier left tail.
- Skewness close to zero suggests approximate symmetry, but the histogram must also be examined.

## Analysis 5: Kurtosis

We will calculate excess kurtosis to understand the tail behavior of the distributions.

We will interpret the result together with the histogram and box plot rather than using kurtosis as a standalone outlier detector.

## Analysis 6: Potential Outliers

We will use box plots and the IQR method to identify potential outliers.

The notebook will flag unusual observations for investigation but will not automatically remove them.

## Expected Outcome

By the end of this analysis, we should be able to explain the shape, center, spread, skewness, and potential outliers of each numerical feature.

We will use the actual statistics and plots generated by the dataset to draw conclusions rather than assuming the results in advance.

```python
# Step 3: Implement the Concept – Distribution Analysis

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Load the Telco Customer Churn dataset
file_path = Path("dataset/Telco-Customer-Churn.csv")

df = pd.read_csv(file_path)

# Convert TotalCharges to numeric
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"],
    errors="coerce"
)

# Select important numerical features
numerical_features = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

print("Dataset shape:", df.shape)
print("\nNumerical features:", numerical_features)


# 1. Calculate Descriptive Statistics
print("\n" + "=" * 60)
print("1. DESCRIPTIVE STATISTICS")
print("=" * 60)

for feature in numerical_features:

    values = df[feature].dropna()

    if values.empty:
        print(f"\nNo valid numerical values for {feature}.")
        continue

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)

    statistics = {
        "Count": values.count(),
        "Mean": values.mean(),
        "Median": values.median(),
        "Mode": values.mode().iloc[0],
        "Minimum": values.min(),
        "Maximum": values.max(),
        "Range": values.max() - values.min(),
        "Variance": values.var(),
        "Standard Deviation": values.std(),
        "Q1": q1,
        "Q3": q3,
        "IQR": q3 - q1,
        "Skewness": values.skew(),
        "Excess Kurtosis": values.kurt()
    }

    print(f"\nFeature: {feature}")

    for name, value in statistics.items():
        if name == "Count":
            print(f"{name}: {value}")
        else:
            print(f"{name}: {value:.3f}")


# 2. Plot Histograms to Understand Distribution Shape
print("\n" + "=" * 60)
print("2. HISTOGRAMS")
print("=" * 60)

for feature in numerical_features:

    values = df[feature].dropna()

    if values.empty:
        continue

    plt.figure(figsize=(8, 5))

    plt.hist(
        values,
        bins=30,
        edgecolor="black"
    )

    plt.title(f"Distribution of {feature}")
    plt.xlabel(feature)
    plt.ylabel("Frequency")
    plt.tight_layout()
    plt.show()


# 3. Plot Box Plots to Inspect Potential Outliers
print("\n" + "=" * 60)
print("3. BOXPLOTS")
print("=" * 60)

for feature in numerical_features:

    values = df[feature].dropna()

    if values.empty:
        continue

    plt.figure(figsize=(8, 4))

    plt.boxplot(
        values,
        vert=False
    )

    plt.title(f"Box Plot of {feature}")
    plt.xlabel(feature)
    plt.tight_layout()
    plt.show()


# 4. Analyze Skewness and Kurtosis
print("\n" + "=" * 60)
print("4. SKEWNESS AND KURTOSIS")
print("=" * 60)

for feature in numerical_features:

    values = df[feature].dropna()

    if values.empty:
        continue

    skewness = values.skew()
    kurtosis = values.kurt()

    if skewness > 0.5:
        skew_interpretation = "Right-skewed"
    elif skewness < -0.5:
        skew_interpretation = "Left-skewed"
    else:
        skew_interpretation = "Approximately symmetric by this rule"

    if kurtosis > 0:
        kurtosis_interpretation = (
            "Positive excess kurtosis; investigate tail behavior"
        )
    elif kurtosis < 0:
        kurtosis_interpretation = (
            "Negative excess kurtosis relative to normal"
        )
    else:
        kurtosis_interpretation = (
            "Excess kurtosis is approximately zero"
        )

    print(f"\nFeature: {feature}")
    print(f"Skewness: {skewness:.3f}")
    print(f"Interpretation: {skew_interpretation}")
    print(f"Excess Kurtosis: {kurtosis:.3f}")
    print(f"Interpretation: {kurtosis_interpretation}")


# 5. Detect Potential Outliers Using IQR
print("\n" + "=" * 60)
print("5. POTENTIAL OUTLIERS USING IQR")
print("=" * 60)

for feature in numerical_features:

    values = df[feature].dropna()

    if values.empty:
        continue

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = values[
        (values < lower_bound)
        | (values > upper_bound)
    ]

    outlier_percentage = (
        len(outliers) / len(values) * 100
    )

    print(f"\nFeature: {feature}")
    print(f"Lower Bound: {lower_bound:.3f}")
    print(f"Upper Bound: {upper_bound:.3f}")
    print(f"Potential Outliers: {len(outliers)}")
    print(f"Percentage Flagged: {outlier_percentage:.2f}%")

    if len(outliers) > 0:
        print("Example potential outlier values:")
        print(outliers.head(10).to_list())


# 6. Compare Mean and Median to Understand the Center
print("\n" + "=" * 60)
print("6. CENTER COMPARISON")
print("=" * 60)

for feature in numerical_features:

    values = df[feature].dropna()

    if values.empty:
        continue

    mean_value = values.mean()
    median_value = values.median()

    print(f"\nFeature: {feature}")
    print(f"Mean: {mean_value:.3f}")
    print(f"Median: {median_value:.3f}")

    if mean_value > median_value:
        print(
            "The mean is greater than the median. "
            "This can occur in a right-skewed distribution."
        )
    elif mean_value < median_value:
        print(
            "The mean is less than the median. "
            "This can occur in a left-skewed distribution."
        )
    else:
        print(
            "The mean and median are equal or very close."
        )


# 7. Final Summary
print("\n" + "=" * 60)
print("7. FINAL SUMMARY")
print("=" * 60)

print("Distribution analysis completed.")
print("Features analyzed:", numerical_features)
print("Histograms and box plots were generated.")
print("Center, spread, skewness, and excess kurtosis were calculated.")
print("Potential outliers were flagged using the IQR method.")
print("No observations were automatically removed or transformed.")
```
