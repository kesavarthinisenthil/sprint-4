# Categorical Data Analysis

## 1. What is Categorical Data Analysis?
Categorical Data Analysis is the process of analyzing data that represents groups, labels, or categories.

Examples:
- Gender: Male, Female
- Contract: Month-to-month, One year, Two year
- InternetService: DSL, Fiber optic, No
- Churn: Yes, No

In AI/ML, categorical analysis helps us understand customer groups and identify patterns that may influence predictions.

## 2. Unique Categories
Unique categories are the different values present in a categorical column.

Example:
The Contract column may contain:
- Month-to-month
- One year
- Two year

We use `unique()` to find the unique categories and `nunique()` to count them.

## 3. Category Frequencies
Category frequency tells us how many times each category appears in a column.

Example:
- Month-to-month: 3,875 customers
- One year: 1,473 customers
- Two year: 1,695 customers

We use `value_counts()` to calculate category frequencies.

## 4. Category Distribution
Category distribution shows how the total records are divided among different categories.

We can calculate the percentage of each category using:

`value_counts(normalize=True) * 100`

This helps us understand which categories are common and which are less common.

## 5. Rare Categories
Rare categories are categories that appear only a few times compared with other categories.

Rare categories may affect model performance, especially when the model does not have enough examples to learn their patterns.

We can identify rare categories by checking their frequency or percentage.

## 6. High Cardinality
High cardinality means a categorical column contains many unique values.

Example:
- CustomerID
- ZIP code
- Product ID

High-cardinality columns may increase the number of features created during encoding. Identifier columns may also provide little useful information for prediction.

## 7. Low Cardinality
Low cardinality means a categorical column contains only a small number of unique values.

Examples:
- Gender
- SeniorCitizen
- Churn

Low-cardinality columns are generally easier to analyze and encode.

## 8. Category Imbalance
Category imbalance occurs when some categories have significantly more records than others.

Example:
- Churn = No: 5,174 customers
- Churn = Yes: 1,869 customers

The target variable is imbalanced because one class has considerably more observations than the other.

Class imbalance may cause a machine learning model to favor the majority class.

## 9. Relationship Between Categorical Features and Target
We can compare categorical features with the target variable to understand how different groups relate to the prediction outcome.

Example:
- Contract vs Churn
- InternetService vs Churn
- PaymentMethod vs Churn

A cross-tabulation helps us compare category counts. Row percentages help us compare the proportion of customers who churn within each group.

## 10. Why Is Categorical Analysis Important in AI/ML?
- Helps identify common and rare categories.
- Helps detect high-cardinality columns.
- Helps identify imbalanced target classes.
- Helps understand relationships between features and the target.
- Supports decisions about categorical encoding and feature selection.
- Helps us prepare data for machine learning models.

Important: A relationship between a category and the target does not automatically prove that the category causes the outcome.

# Categorical Data Analysis

## Business Use Case: Telecom Customer Churn

In this notebook, we will analyze categorical columns in the Telco Customer Churn dataset.

Dataset path: `dataset/Telco-Customer-Churn.csv`

The dataset contains information about telecom customers, including their contract type, internet service, payment method, and churn status.

### Objectives

1. Identify categorical columns in the dataset.
2. Display the unique categories in important columns.
3. Calculate category frequencies and percentages.
4. Identify rare categories.
5. Compare high-cardinality and low-cardinality columns.
6. Analyze the distribution of the Churn target variable.
7. Compare Contract, InternetService, and PaymentMethod with Churn.
8. Use cross-tabulations and row percentages to understand customer churn patterns.
9. Visualize categorical distributions using bar charts.

### Expected Business Insights

- Identify the most common contract type.
- Understand which payment methods are frequently used.
- Check whether the target variable is imbalanced.
- Compare churn percentages across contract types.
- Understand how churn varies across internet services and payment methods.
- Identify categorical columns that may require special handling before machine learning.

We will analyze the data without removing records or modifying the original dataset.

```python
# Step 3: Implement Concept – Categorical Data Analysis

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# --------------------------------------------------
# 1. Load the Dataset
# --------------------------------------------------

file_path = Path("dataset/Telco-Customer-Churn.csv")

if not file_path.exists():
    raise FileNotFoundError(
        f"Dataset not found at: {file_path.resolve()}\n"
        "Make sure the dataset folder is in the same directory as this notebook."
    )

df = pd.read_csv(file_path)

print("DATASET LOADED SUCCESSFULLY")
print("-" * 60)
print("Dataset shape:", df.shape)
print("\nFirst five rows:")
display(df.head())


# --------------------------------------------------
# 2. Identify Categorical Columns
# --------------------------------------------------

categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("\nCATEGORICAL COLUMNS")
print("-" * 60)

for column in categorical_columns:
    print(column)

print("\nTotal categorical columns:", len(categorical_columns))


# --------------------------------------------------
# 3. Display Unique Categories
# --------------------------------------------------

print("\nUNIQUE CATEGORIES")
print("-" * 60)

for column in categorical_columns:
    print(f"\nColumn: {column}")
    print("Number of unique categories:", df[column].nunique(dropna=False))
    print("Unique values:", df[column].unique())


# --------------------------------------------------
# 4. Calculate Category Frequencies and Percentages
# --------------------------------------------------

print("\nCATEGORY FREQUENCIES AND PERCENTAGES")
print("-" * 60)

for column in categorical_columns:
    print(f"\nColumn: {column}")

    frequency = df[column].value_counts(dropna=False)
    percentage = (
        df[column].value_counts(normalize=True, dropna=False) * 100
    )

    summary = pd.DataFrame({
        "Frequency": frequency,
        "Percentage": percentage.round(2)
    })

    display(summary)


# --------------------------------------------------
# 5. Identify Rare Categories
# --------------------------------------------------

print("\nRARE CATEGORY ANALYSIS")
print("-" * 60)

# A category is considered rare here if it represents
# less than 1% of the records in its column.
# This is an example threshold, not a universal rule.

rare_threshold = 1.0

for column in categorical_columns:
    percentages = (
        df[column].value_counts(normalize=True, dropna=False) * 100
    )

    rare_categories = percentages[percentages < rare_threshold]

    print(f"\nColumn: {column}")

    if rare_categories.empty:
        print("No categories below the 1% threshold.")
    else:
        print("Categories below the 1% threshold:")
        display(rare_categories.rename("Percentage").to_frame().round(2))


# --------------------------------------------------
# 6. Identify High- and Low-Cardinality Columns
# --------------------------------------------------

print("\nCARDINALITY ANALYSIS")
print("-" * 60)

cardinality_summary = pd.DataFrame({
    "Column": categorical_columns,
    "Unique_Categories": [
        df[column].nunique(dropna=False)
        for column in categorical_columns
    ]
})

# These thresholds are illustrative and can be adjusted.
cardinality_summary["Cardinality_Type"] = np.where(
    cardinality_summary["Unique_Categories"] <= 10,
    "Low Cardinality",
    "High Cardinality"
)

display(
    cardinality_summary.sort_values(
        by="Unique_Categories",
        ascending=False
    ).reset_index(drop=True)
)

print(
    "\nNote: The threshold of 10 unique categories is an "
    "illustrative rule for this analysis."
)


# --------------------------------------------------
# 7. Analyze Target Variable Distribution
# --------------------------------------------------

target_column = "Churn"

if target_column in df.columns:
    print("\nTARGET VARIABLE DISTRIBUTION")
    print("-" * 60)

    churn_counts = df[target_column].value_counts(dropna=False)
    churn_percentages = (
        df[target_column].value_counts(normalize=True, dropna=False) * 100
    )

    churn_summary = pd.DataFrame({
        "Frequency": churn_counts,
        "Percentage": churn_percentages.round(2)
    })

    display(churn_summary)

    plt.figure(figsize=(7, 5))
    churn_counts.plot(kind="bar")

    plt.title("Distribution of Customer Churn")
    plt.xlabel("Churn Status")
    plt.ylabel("Number of Customers")
    plt.xticks(rotation=0)
    plt.tight_layout()
    plt.show()

    if len(churn_counts) > 1:
        imbalance_ratio = churn_counts.max() / churn_counts.min()

        print("Majority-to-minority class ratio:", round(imbalance_ratio, 2))

        if imbalance_ratio >= 1.5:
            print("Observation: The target classes show an imbalance.")
        else:
            print("Observation: The target classes are relatively balanced.")

else:
    print(f"Target column '{target_column}' was not found.")


# --------------------------------------------------
# 8. Visualize Important Categorical Distributions
# --------------------------------------------------

important_columns = [
    "Contract",
    "InternetService",
    "PaymentMethod"
]

print("\nIMPORTANT CATEGORICAL DISTRIBUTIONS")
print("-" * 60)

available_columns = [
    column for column in important_columns
    if column in df.columns
]

for column in available_columns:
    counts = df[column].value_counts(dropna=False)

    print(f"\nCategory frequencies for {column}:")
    display(counts.rename("Frequency").to_frame())

    plt.figure(figsize=(9, 5))
    counts.plot(kind="bar")

    plt.title(f"Category Distribution: {column}")
    plt.xlabel(column)
    plt.ylabel("Number of Customers")
    plt.xticks(rotation=35, ha="right")
    plt.tight_layout()
    plt.show()


# --------------------------------------------------
# 9. Analyze Contract vs Churn
# --------------------------------------------------

print("\nCONTRACT VS CHURN")
print("-" * 60)

if "Contract" in df.columns and target_column in df.columns:

    contract_counts = pd.crosstab(
        df["Contract"],
        df[target_column],
        dropna=False
    )

    print("Customer counts:")
    display(contract_counts)

    contract_percentages = pd.crosstab(
        df["Contract"],
        df[target_column],
        normalize="index",
        dropna=False
    ) * 100

    print("\nChurn percentages within each contract type:")
    display(contract_percentages.round(2))

    if "Yes" in contract_percentages.columns:
        print("\nChurn percentage by contract type:")

        for contract, percentage in contract_percentages["Yes"].items():
            print(f"{contract}: {percentage:.2f}%")

    contract_counts.plot(kind="bar", figsize=(9, 5))

    plt.title("Contract Type vs Churn")
    plt.xlabel("Contract Type")
    plt.ylabel("Number of Customers")
    plt.xticks(rotation=20, ha="right")
    plt.tight_layout()
    plt.show()

else:
    print("Contract or Churn column is unavailable.")


# --------------------------------------------------
# 10. Analyze InternetService vs Churn
# --------------------------------------------------

print("\nINTERNET SERVICE VS CHURN")
print("-" * 60)

if "InternetService" in df.columns and target_column in df.columns:

    internet_counts = pd.crosstab(
        df["InternetService"],
        df[target_column],
        dropna=False
    )

    print("Customer counts:")
    display(internet_counts)

    internet_percentages = pd.crosstab(
        df["InternetService"],
        df[target_column],
        normalize="index",
        dropna=False
    ) * 100

    print("\nChurn percentages within each internet service group:")
    display(internet_percentages.round(2))

    if "Yes" in internet_percentages.columns:
        print("\nChurn percentage by internet service:")

        for service, percentage in internet_percentages["Yes"].items():
            print(f"{service}: {percentage:.2f}%")

    internet_counts.plot(kind="bar", figsize=(9, 5))

    plt.title("Internet Service vs Churn")
    plt.xlabel("Internet Service")
    plt.ylabel("Number of Customers")
    plt.xticks(rotation=20, ha="right")
    plt.tight_layout()
    plt.show()

else:
    print("InternetService or Churn column is unavailable.")


# --------------------------------------------------
# 11. Analyze PaymentMethod vs Churn
# --------------------------------------------------

print("\nPAYMENT METHOD VS CHURN")
print("-" * 60)

if "PaymentMethod" in df.columns and target_column in df.columns:

    payment_counts = pd.crosstab(
        df["PaymentMethod"],
        df[target_column],
        dropna=False
    )

    print("Customer counts:")
    display(payment_counts)

    payment_percentages = pd.crosstab(
        df["PaymentMethod"],
        df[target_column],
        normalize="index",
        dropna=False
    ) * 100

    print("\nChurn percentages within each payment method:")
    display(payment_percentages.round(2))

    if "Yes" in payment_percentages.columns:
        print("\nChurn percentage by payment method:")

        for method, percentage in payment_percentages["Yes"].items():
            print(f"{method}: {percentage:.2f}%")

    payment_counts.plot(kind="bar", figsize=(10, 5))

    plt.title("Payment Method vs Churn")
    plt.xlabel("Payment Method")
    plt.ylabel("Number of Customers")
    plt.xticks(rotation=30, ha="right")
    plt.tight_layout()
    plt.show()

else:
    print("PaymentMethod or Churn column is unavailable.")


# --------------------------------------------------
# 12. Final Summary
# --------------------------------------------------

print("\nFINAL SUMMARY")
print("-" * 60)

print("1. Identified categorical columns and their unique values.")
print("2. Calculated category frequencies and percentages.")
print("3. Identified categories below the selected rare-category threshold.")
print("4. Classified categorical columns by cardinality.")
print("5. Examined the target variable distribution.")
print("6. Compared Contract, InternetService, and PaymentMethod with Churn.")
print("7. Visualized categorical distributions and group comparisons.")

print(
    "\nBusiness note: Compare churn percentages within each category "
    "rather than only comparing raw customer counts."
)

print(
    "These relationships describe patterns in the dataset; "
    "they do not prove that a category causes customer churn."
)

print("\nCategorical data analysis completed successfully.")
```
