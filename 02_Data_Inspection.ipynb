# Notebook 2 – Data Inspection

## Step 1: Understand Concept

### 1. What is Data Inspection?

Data inspection is the process of examining a dataset to understand its structure, size, columns, and data types before performing data analysis.

### 2. Why Do We Need Data Inspection?

* To understand the number of rows and columns.
* To identify the data types of each column.
* To check whether the dataset contains numerical or categorical data.
* To understand the values present in each column.
* To identify columns that may need further data-quality checks.

### 3. Important Pandas Functions

| Function         | Purpose                                                                         |
| ---------------- | ------------------------------------------------------------------------------- |
| `head()`         | Displays the first five rows by default.                                        |
| `tail()`         | Displays the last five rows by default.                                         |
| `sample()`       | Displays randomly selected rows.                                                |
| `shape`          | Returns the number of rows and columns.                                         |
| `info()`         | Displays column names, data types, and non-null counts.                         |
| `describe()`     | Generates descriptive statistics for numerical columns by default.              |
| `dtypes`         | Displays the data type of each column.                                          |
| `nunique()`      | Counts the distinct values in each column, excluding missing values by default. |
| `value_counts()` | Counts how often each value appears in a column.                                |

### 4. Types of Columns

* **Numerical columns:** Contain numbers, such as tenure and monthly charges.
* **Categorical columns:** Contain categories, such as contract type and payment method.
* **Date/Time columns:** Contain dates or times.
* **Identifier columns:** Identify individual records, such as customer ID.
* **Potential target column:** The column we want to predict, such as `Churn`.

### 5. Dataset

We will inspect the Telco Customer Churn dataset to understand its structure before performing Exploratory Data Analysis (EDA).


## Step 2: Demonstrate Concept

### Real-World Example: Customer Churn Analysis

A telecommunications company wants to understand why customers leave its service.

Before analyzing customer churn, we need to inspect the dataset.

For example:

| Customer ID | Tenure | Monthly Charges | Contract       | Churn |
| ----------- | -----: | --------------: | -------------- | ----- |
| 001         |     12 |            70.5 | Month-to-month | Yes   |
| 002         |     36 |            85.0 | Two year       | No    |
| 003         |      5 |            55.2 | Month-to-month | Yes   |

*Note: This table is illustrative and does not represent actual dataset records.*

Using Pandas inspection functions, we can:

* View the first and last customer records.
* Check the number of customers and columns.
* Identify the data type of each column.
* Examine the distribution of values in the `Churn` column.
* Identify columns that may require additional investigation.

**Expected outcome:** We will understand the dataset's structure and know which columns to investigate further during EDA.


```python
# Import the required library
import pandas as pd
from pathlib import Path

# Load the Telco Customer Churn dataset
file_path = Path("dataset/Telco-Customer-Churn.csv")
df = pd.read_csv(file_path)

# 1. Display the first five rows
print("First Five Rows:")
display(df.head())

# 2. Display the last five rows
print("\nLast Five Rows:")
display(df.tail())

# 3. Display five random rows
print("\nRandom Five Rows:")
display(df.sample(5, random_state=42))

# 4. Display the number of rows and columns
print("\nDataset Shape:")
print(df.shape)

# 5. Display dataset information
print("\nDataset Information:")
df.info()

# 6. Display descriptive statistics
print("\nDescriptive Statistics:")
display(df.describe(include="all").T)

# 7. Display the data type of each column
print("\nColumn Data Types:")
print(df.dtypes)

# 8. Count unique values in each column
print("\nUnique Values Per Column:")
print(df.nunique())

# 9. Display the number of unique values in the Churn column
print("\nChurn Value Counts:")
print(df["Churn"].value_counts(dropna=False))

# 10. Classify columns by data type
numerical_columns = df.select_dtypes(include="number").columns.tolist()

categorical_columns = df.select_dtypes(
    include=["object", "category", "bool"]
).columns.tolist()

print("\nNumerical Columns:")
print(numerical_columns)

print("\nCategorical Columns:")
print(categorical_columns)

# 11. Identify the customer identifier
print("\nIdentifier Column: customerID")

# 12. Identify the target variable
print("Target Variable: Churn")
```
