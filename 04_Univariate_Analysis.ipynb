# Notebook 4 – Univariate Analysis

### 1. What Is Univariate Analysis?

Univariate analysis is the process of analyzing one variable at a time to understand its distribution, central tendency, spread, and frequency.

It helps us understand individual features before exploring relationships between variables.

### 2. Numerical Variable Analysis

Numerical variables contain measurable or countable values. In our Telco Customer Churn dataset, examples include `tenure`, `MonthlyCharges`, and `TotalCharges`.

| Statistical Measure | Purpose                                                                                                |
| ------------------- | ------------------------------------------------------------------------------------------------------ |
| Mean                | Calculates the average value.                                                                          |
| Median              | Finds the middle value after sorting.                                                                  |
| Mode                | Finds the most frequently occurring value.                                                             |
| Minimum             | Finds the smallest value.                                                                              |
| Maximum             | Finds the largest value.                                                                               |
| Range               | Calculates maximum minus minimum.                                                                      |
| Variance            | Measures the spread of values around the mean using squared deviations.                                |
| Standard Deviation  | Measures the typical spread around the mean, in the original units.                                    |
| Quartiles           | Divide ordered data into four parts.                                                                   |
| Percentiles         | Show the value below which a given percentage of observations falls.                                   |
| Skewness            | Measures the asymmetry of a distribution.                                                              |
| Kurtosis            | Describes tail heaviness and the tendency to produce extreme values relative to a normal distribution. |

### 3. Numerical Visualizations

* **Histogram:** Shows the frequency distribution of numerical values.
* **KDE / Density Plot:** Shows a smoothed estimate of a distribution.
* **Box Plot:** Displays the median, quartiles, spread, and potential outliers.

### 4. Categorical Variable Analysis

Categorical variables represent groups or labels. Examples include `Contract`, `PaymentMethod`, `InternetService`, and `Churn`.

| Measure                 | Purpose                                                          |
| ----------------------- | ---------------------------------------------------------------- |
| Unique Values           | Lists the different categories present.                          |
| Frequency               | Counts observations in each category.                            |
| Value Counts            | Shows the number of observations per category.                   |
| Percentage Distribution | Shows the percentage of observations belonging to each category. |

### 5. Categorical Visualizations

* **Bar Chart:** Compares the frequency of different categories.
* **Count Plot:** Displays the number of observations in each category.
* **Pie Chart:** Shows proportions of categories when there are only a few categories.

### 6. Why Is Univariate Analysis Important in AI/ML?

Univariate analysis helps us understand feature distributions, detect potential outliers, identify skewed data, and recognize imbalanced target classes before building machine learning models.

In this notebook, we will analyze customer tenure, monthly charges, total charges, and important categorical features in the Telco Customer Churn dataset.


### Real-World Example: Understanding Customer Churn

A telecommunications company wants to understand its customers before building a churn prediction model.

### Numerical Analysis

**Example: Customer Tenure**

Tenure represents the number of months a customer has stayed with the company.

By calculating the mean and median, we can understand the typical customer tenure. The minimum, maximum, and range show the overall spread.

A histogram helps us identify where most customers fall, while a box plot helps us identify unusually low or high tenure values.

**Example: Monthly Charges**

Monthly charges represent the amount a customer pays each month.

Mean, median, standard deviation, skewness, and a histogram help us understand typical charges and how widely they vary.

### Categorical Analysis

**Example: Contract**

By examining the `Contract` column, we can identify how many customers have month-to-month, one-year, and two-year contracts.

A bar chart helps us compare the number of customers in each contract category.

**Example: Churn**

The `Churn` column indicates whether a customer left the company.

Its frequency and percentage distribution show the balance between customers who left and those who stayed.

### How We Interpret the Findings

* If the mean and median are different, the distribution may be skewed.
* A large standard deviation indicates greater variation in numerical values.
* A box plot may reveal potential outliers that need further investigation.
* A category with a high percentage represents a large share of the dataset.
* If one churn category dominates, the target may be imbalanced.

**Important:** These are interpretation guidelines. The actual findings must be based on the results produced by our dataset.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

# Load the dataset
file_path = Path("dataset/Telco-Customer-Churn.csv")
df = pd.read_csv(file_path)

# Convert TotalCharges to numeric.
# Some values may contain blank spaces and become NaN.
df["TotalCharges"] = pd.to_numeric(
    df["TotalCharges"], errors="coerce"
)

# Select important numerical and categorical features
numerical_columns = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]

categorical_columns = [
    "Contract",
    "PaymentMethod",
    "InternetService",
    "Churn"
]

# --------------------------------------------------
# 1. NUMERICAL VARIABLE ANALYSIS
# --------------------------------------------------

print("=" * 60)
print("NUMERICAL VARIABLE ANALYSIS")
print("=" * 60)

for column in numerical_columns:

    data = df[column].dropna()

    if data.empty:
        print(f"\nNo valid numerical data in {column}.")
        continue

    print(f"\n{'-' * 50}")
    print(f"Feature: {column}")
    print(f"{'-' * 50}")

    # Central tendency
    print("\nCentral Tendency:")
    print("Mean:", round(data.mean(), 2))
    print("Median:", round(data.median(), 2))
    print("Mode:", data.mode().tolist())

    # Minimum, maximum, and range
    print("\nSpread:")
    print("Minimum:", round(data.min(), 2))
    print("Maximum:", round(data.max(), 2))
    print("Range:", round(data.max() - data.min(), 2))
    print("Variance:", round(data.var(), 2))
    print("Standard Deviation:", round(data.std(), 2))

    # Quartiles
    print("\nQuartiles:")
    print(data.quantile([0.25, 0.50, 0.75]))

    # Percentiles
    print("\nPercentiles:")
    print(data.quantile([0.10, 0.25, 0.50, 0.75, 0.90]))

    # Skewness and kurtosis
    print("\nDistribution Shape:")
    print("Skewness:", round(data.skew(), 2))
    print("Kurtosis:", round(data.kurt(), 2))

    # Automatic interpretation
    print("\nFindings:")

    if data.mean() > data.median():
        print("- Mean is greater than median; this may indicate right skew.")
    elif data.mean() < data.median():
        print("- Mean is less than median; this may indicate left skew.")
    else:
        print("- Mean and median are approximately equal.")

    skewness = data.skew()

    if skewness > 0.5:
        print("- The distribution shows noticeable right skew.")
    elif skewness < -0.5:
        print("- The distribution shows noticeable left skew.")
    else:
        print("- The distribution is approximately symmetric by skewness.")

    print(
        "- Standard deviation describes how widely values vary "
        "around the mean."
    )

    # Visualizations: Histogram, KDE, and Box Plot
    fig, axes = plt.subplots(1, 3, figsize=(18, 4))

    axes[0].hist(data, bins=30, edgecolor="black")
    axes[0].set_title(f"Histogram of {column}")
    axes[0].set_xlabel(column)
    axes[0].set_ylabel("Frequency")

    # KDE requires SciPy; skip gracefully if unavailable
    try:
        data.plot(kind="kde", ax=axes[1])
        axes[1].set_title(f"KDE Plot of {column}")
        axes[1].set_xlabel(column)
    except (ImportError, ModuleNotFoundError):
        axes[1].text(
            0.5, 0.5,
            "KDE unavailable\nInstall SciPy to enable",
            ha="center", va="center"
        )
        axes[1].set_title(f"KDE Plot of {column}")

    axes[2].boxplot(data, vert=True)
    axes[2].set_title(f"Box Plot of {column}")
    axes[2].set_ylabel(column)

    plt.tight_layout()
    plt.show()


# --------------------------------------------------
# 2. CATEGORICAL VARIABLE ANALYSIS
# --------------------------------------------------

print("\n" + "=" * 60)
print("CATEGORICAL VARIABLE ANALYSIS")
print("=" * 60)

for column in categorical_columns:

    data = df[column].dropna()

    if data.empty:
        print(f"\nNo valid categorical data in {column}.")
        continue

    print(f"\n{'-' * 50}")
    print(f"Feature: {column}")
    print(f"{'-' * 50}")

    # Unique values
    print("\nUnique Values:")
    print(data.unique())

    # Frequency and percentage distribution
    frequency = data.value_counts()
    percentage = (
        data.value_counts(normalize=True) * 100
    ).round(2)

    distribution = pd.DataFrame({
        "Frequency": frequency,
        "Percentage": percentage
    })

    print("\nFrequency and Percentage Distribution:")
    display(distribution)

    # Automatic interpretation
    highest_category = frequency.idxmax()
    highest_percentage = percentage.loc[highest_category]

    print("\nFindings:")
    print(
        f"- The most frequent category is '{highest_category}', "
        f"representing {highest_percentage:.2f}% of non-missing values."
    )
    print(f"- Number of unique categories: {data.nunique()}")

    # Bar chart / count plot
    plt.figure(figsize=(9, 4))
    frequency.plot(kind="bar", edgecolor="black")
    plt.title(f"Category Frequencies: {column}")
    plt.xlabel(column)
    plt.ylabel("Frequency")
    plt.xticks(rotation=30, ha="right")
    plt.tight_layout()
    plt.show()

    # Pie chart for variables with few categories
    if data.nunique() <= 5:
        plt.figure(figsize=(6, 6))
        frequency.plot(
            kind="pie",
            autopct="%1.1f%%",
            startangle=90
        )
        plt.title(f"Percentage Distribution: {column}")
        plt.ylabel("")
        plt.tight_layout()
        plt.show()


# --------------------------------------------------
# 3. FINAL SUMMARY
# --------------------------------------------------

print("\n" + "=" * 60)
print("FINAL SUMMARY")
print("=" * 60)

print("Dataset dimensions:", df.shape)
print("\nMissing values in selected numerical features:")
print(df[numerical_columns].isnull().sum())

print("\nMissing values in selected categorical features:")
print(df[categorical_columns].isnull().sum())

print(
    "\nUnivariate analysis completed. Review the statistics "
    "and charts to understand each feature's distribution."
)
```
